İşin Matematiği: Kısmi TürevlerDenklemimiz artık şu şekle geldi:$$y = w \cdot x + b$$Modelin hata fonksiyonu (Loss) ise hala aynı: Tahmin ile gerçek değerin farkının karesi.$$Loss = (y_{tahmin} - y)^2$$

Artık elimizde optimize edilecek iki değişken ($w$ ve $b$) olduğu için, hatayı her ikisine göre ayrı ayrı küçültmemiz gerekiyor. Bunun için hata fonksiyonunun $w$'ya ve $b$'ye göre kısmi türevlerini (partial derivatives) alırız:

$w$'ya göre türev: Hatayı zincir kuralıyla (chain rule) açtığımızda, $x$ çarpım durumunda olduğu için dışarı çıkar.$$\frac{\partial Loss}{\partial w} = 2 \cdot x \cdot (y_{tahmin} - y)$$$b$'ye göre türev: $b$'nin yanında çarpım durumunda bir $x$ olmadığı için (aslında katsayısı 1'dir), türev sadece hatanın kendisine bağlı kalır.$$\frac{\partial Loss}{\partial b} = 2 \cdot (y_{tahmin} - y)$$

```python
# 1. VERİ SETİMİZ
# x_data: Oda sayısı, y_data: Evin fiyatı (1000 TL cinsinden)
# Gerçek kuralımız: Fiyat = 10 * Oda + 5 (Yani taban fiyat 5 bin TL)
x_data = [1.0, 2.0, 3.0, 4.0]
y_data = [15.0, 25.0, 35.0, 45.0]
x_data = [(x - min(x_data)) / (max(x_data) - min(x_data)) for x in x_data]
```

```python
# 2. BAŞLANGIÇ DURUMU
w = 0.0  # Ağırlık (Weight)
b = 0.0  # Sapma (Bias)
```

```python
w = 0.0
b = 0.0
learning_rate = 0.01   # toplu güncellemede daha büyük kullanılabilir
```

```python
for epoch in range(1, 101):
    toplam_hata = 0.0
    dw = 0.0   # gradyan biriktirici (w için)
    db = 0.0   # gradyan biriktirici (b için)

    for x, y in zip(x_data, y_data):
        y_tahmin = w * x + b
        hata = y_tahmin - y
        toplam_hata += hata**2

        # Gradyanları topla (toplu güncelleme için)
        dw += 2 * x * hata
        db += 2 * hata

    # Ortalama gradyan (veri sayısına böl)
    N = len(x_data)
    dw /= N
    db /= N

    # Ağırlıkları güncelle
    w -= learning_rate * dw
    b -= learning_rate * db

    if epoch % 20 == 0:
        print(f"Adım {epoch} | w: {w:.4f} | b: {b:.4f} | Toplam Hata: {toplam_hata:.4f}")
```

```python
print("\nEğitim Bitti!")
print(f"Bilgisayarın bulduğu kural: y = {w:.2f} * x + {b:.2f}")
```

```python
# 4. YENİ BİR VERİ İLE TEST
yeni_oda_sayisi = 5.0
tahmin = w * yeni_oda_sayisi + b
print(f"\nModelin Tahmini: 5 odalı bir ev {tahmin:.2f} bin TL eder.")
```
